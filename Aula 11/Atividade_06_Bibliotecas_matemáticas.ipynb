#  **Gram-Schmidt: Explicação e Atividade**
 **Teoria: Processo de Gram-Schmidt**
 O processo de Gram-Schmidt é um método para transformar um conjunto de vetores linearmente independentes em um conjunto ortogonal ou ortonormal. Ele é amplamente usado em álgebra linear para problemas de diagonalização de matrizes, espaços vetoriais e decomposição QR.

# #### Fórmulas:
 1. **Primeiro vetor ortogonal:**
 $$u_1 = v_1$$

# 2. **Ortogonalização dos vetores subsequentes:**
 $$u_i = v_i - \sum_{j=1}^{i-1} \text{proj}_{u_j}(v_i)$$
 Onde:
 $$\text{proj}_{u_j}(v_i) = \frac{\langle v_i, u_j \rangle}{\langle u_j, u_j \rangle} u_j$$

 3. **Ortonormalização (opcional):**
 $$e_i = \frac{u_i}{\|u_i\|}$$

 # Propriedades:
 - Os vetores \(u_1, u_2, \dots, u_n\) formam um conjunto ortogonal.
 - Se normalizados, formam um conjunto ortonormal.

````python
# Importação das bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar o dataset
url = "/content/supermarket_sales new.csv"
df = pd.read_csv(url)

# Exibir as primeiras informações do dataset
print("Primeiras linhas do dataset:")
print(df.head())

# -------- Atividade 1: Limpeza de Dados --------
print("\n1. Existem valores ausentes no dataset?")
# Mostrar a quantidade de valores ausentes por coluna
valores_ausentes = df.isnull().sum()
print(valores_ausentes)

# -------- Atividade 2: Perguntas Analíticas --------

# Pergunta 1: Qual é a distribuição das vendas entre as cidades?
print("\nPergunta 1: Qual é a distribuição das vendas entre as cidades?")
# Contagem de vendas por cidade
vendas_por_cidade = df.groupby('City').size()

# Gráfico de barras para visualização
plt.figure(figsize=(8, 6))
vendas_por_cidade.plot(kind='bar', color='skyblue')
plt.title("Distribuição de Vendas por Cidade", fontsize=14)
plt.xlabel('Cidade', fontsize=12)
plt.ylabel('Número de Vendas', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Pergunta 2: Qual é a média do Total de vendas por linha de produtos?
print("\nPergunta 2: Qual é a média do Total de vendas por linha de produtos?")
# Calcular o total de vendas (Preço Unitário * Quantidade + Imposto)
df['Total'] = df['Unit price'] * df['Quantity'] + df['Tax 5%']

# Calcular a média do total de vendas por linha de produtos
media_vendas_por_produto = df.groupby('Product line')['Total'].mean()
print(media_vendas_por_produto)

# Pergunta 3: Existe alguma relação entre o tipo de cliente e o valor total das vendas?
print("\nPergunta 3: Existe alguma relação entre o tipo de cliente e o valor total das vendas?")
# Calcular a média de vendas por tipo de cliente
vendas_por_cliente = df.groupby('Customer type')['Total'].mean()

# Gráfico de barras para visualização
plt.figure(figsize=(8, 6))
vendas_por_cliente.plot(kind='bar', color='lightgreen')
plt.title("Média de Vendas por Tipo de Cliente", fontsize=14)
plt.xlabel('Tipo de Cliente', fontsize=12)
plt.ylabel('Média de Vendas (Total)', fontsize=12)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Pergunta 4: Há correlação entre Preço por Unidade e Total de Vendas?
print("\nPergunta 4: Há correlação entre Preço por Unidade e Total de Vendas?")
# Calcular a correlação entre o Preço Unitário e o Total de Vendas
correlacao_preco_total = df['Unit price'].corr(df['Total'])
print(f"Correlação entre Preço Unitário e Total de Vendas: {correlacao_preco_total:.2f}")

# -------- Atividade 3: Cálculos Adicionais --------

# Calcular o gasto médio de 5% de imposto para cada gênero
print("\nAtividade 3: Gasto médio de 5% de imposto por gênero")
gasto_medio_por_genero = df.groupby('Gender')['Tax 5%'].mean()
print(gasto_medio_por_genero)

# Quantidade total vendida para cada linha de produto
print("\nQuantidade total vendida para cada linha de produto")
quantidade_vendida_por_produto = df.groupby('Product line')['Quantity'].sum()
print(quantidade_vendida_por_produto)

# Faturamento total por cidade
print("\nFaturamento total por cidade")
faturamento_por_cidade = df.groupby('City')['Total'].sum()
print(faturamento_por_cidade)

# Encontrar o produto mais vendido para cada tipo de cliente
print("\nProduto mais vendido para cada tipo de cliente")
produto_mais_vendido_por_cliente = df.groupby('Customer type')['Product line'].apply(lambda x: x.value_counts().idxmax())
print(produto_mais_vendido_por_cliente)

# Calcular o desvio padrão dos preços unitários para cada linha de produto
print("\nDesvio padrão dos preços unitários por linha de produto")
desvio_padrao_precos = df.groupby('Product line')['Unit price'].std()
print(desvio_padrao_precos)
````

````python
from sympy import symbols, Eq, solve

# 1. Definindo os símbolos
x, y, z, w, v = symbols('x y z w v')

# 2. Representando as equações iniciais
eq1 = Eq(2*x + 3*y, 6)      # 2x + 3y = 6
eq2 = Eq(4*x - y, 5)        # 4x - y = 5

# 3. Resolva o sistema para x e y
solution_2d = solve([eq1, eq2], (x, y))
print("Solução para x e y:", solution_2d)

# 4. Adicionando a nova equação
eq3 = Eq(x - y + z, 1)      # x - y + z = 1

# 5. Resolva o sistema expandido para x, y e z
solution_3d = solve([eq1, eq2, eq3], (x, y, z))
print("Solução para x, y e z:", solution_3d)

# 6. Adicionando mais duas equações ao sistema
eq4 = Eq(w + x + y, 10)     # w + x + y = 10
eq5 = Eq(v - 2*y + z, 4)    # v - 2y + z = 4

# 7. Resolva o sistema expandido para x, y, z, w e v
solution_5d = solve([eq1, eq2, eq3, eq4, eq5], (x, y, z, w, v))
print("Solução para x, y, z, w e v:", solution_5d)

# 8. Verificando se a solução encontrada satisfaz todas as equações
print("\nVerificando se a solução satisfaz as equações:")

# Substitui as variáveis na equação para verificar
for eq in [eq1, eq2, eq3, eq4, eq5]:
    print(f"Equação {eq}: {eq.subs(solution_5d)}")
````
