# Paradigmas de programação
  * Referem-se a estilos ou abordagens fundamentais para o desenvolvimento de software. Cada paradigma representa uma maneira particular de pensar sobre como resolver problemas computacionais e organizar o código fonte. Cada um tem suas próprias regras, técnicas e melhores práticas que definem como os programas devem ser estruturados e organizados.


## Programação Estruturada

A programação estruturada é um paradigma de programação que se baseia na ideia de que um programa pode ser estruturado de maneira lógica e hierárquica através de estruturas de controle bem definidas, como sequência, seleção e repetição. Este paradigma surgiu como uma resposta à complexidade e à falta de clareza encontradas em programas desenvolvidos de maneira não estruturada.

### Princípios da Programação Estruturada:

1. **Sequência:** As instruções são executadas em ordem sequencial, de cima para baixo, linha por linha.

2. **Seleção (Decisão):** Utilização de estruturas condicionais como `if`, `else if` (em algumas linguagens) e `else` para tomar decisões baseadas em condições.

3. **Repetição (Laços):** Uso de estruturas de repetição como `for` e `while` para repetir um bloco de código enquanto uma condição for verdadeira.

4. **Sub-rotinas:** Divisão do código em pequenas funções ou procedimentos, permitindo modularidade e reutilização de código.


### Vantagens da Programação Estruturada:

- **Legibilidade:** Código estruturado é mais fácil de entender e manter, pois segue uma organização lógica e hierárquica.
  
- **Facilidade de Depuração:** Erros são mais fáceis de encontrar e corrigir devido à estrutura clara e previsível do código.
  
- **Manutenção:** A divisão em blocos estruturados facilita alterações e atualizações no código sem introduzir muitos riscos de erros.
  
- **Portabilidade:** Programas estruturados são mais facilmente portáveis entre diferentes plataformas e ambientes de desenvolvimento.

### Desvantagens da Programação Estruturada:

- **Limitações em Projetos Grandes:** Em projetos muito grandes e complexos, a estruturação estrita pode tornar o código mais verboso e difícil de gerenciar.

- **Dificuldade em Expressar Algoritmos Complexos:** Algoritmos que envolvem múltiplos fluxos de controle ou dependências complexas podem ser mais difíceis de expressar de forma clara e eficiente na programação estruturada.

- **Menor Flexibilidade em Certas Situações:** Em alguns casos, a estruturação rígida pode limitar a criatividade e a flexibilidade na implementação de soluções.

### Pilares da Programação Orientada a Objetos (POO)

1. **Abstração:**
   - Permite representar características essenciais de objetos do mundo real de forma simplificada em software, focando nos detalhes relevantes e ocultando os irrelevantes.

2. **Encapsulamento:**
   - Agrupa dados (atributos) e operações (métodos) relacionados em uma única unidade, chamada de classe. O encapsulamento protege os dados internos da classe e controla como eles são acessados e modificados por código externo.

3. **Herança:**
   - Permite criar uma nova classe (subclasse) a partir de uma classe existente (superclasse), aproveitando seus atributos e métodos. A herança promove a reutilização de código e a extensibilidade do software.

4. **Polimorfismo:**
   - Permite que objetos de diferentes classes sejam tratados de maneira uniforme, facilitando a flexibilidade e extensibilidade do código. O polimorfismo permite que métodos de uma classe base sejam sobrescritos em classes derivadas, adaptando seu comportamento conforme necessário.

### Persistência de Dados em Python
Persistência de dados refere-se à capacidade de armazenar e recuperar informações entre diferentes execuções de um programa. Cada tipo de arquivo tem suas vantagens e é escolhido com base na estrutura dos dados, no desempenho necessário e na interoperabilidade com outros sistemas. Escolher o tipo de arquivo correto é crucial para garantir que os dados sejam armazenados de forma segura e possam ser facilmente recuperados e manipulados conforme necessário
#### 1. Arquivos de Texto (`txt`)

##### Abrir, Ler e Escrever em um Arquivo de Texto:

Os arquivos de texto simples (`txt`) são amplamente utilizados para armazenar dados legíveis por humanos. Eles são úteis para armazenar informações que não precisam de estruturação complexa.

- Escrever em um arquivo de texto (`txt`):
  


```python
with open('dados.txt', 'w') as arquivo:
      arquivo.write('Este é um exemplo de texto.\n')
      arquivo.write('Python é uma linguagem poderosa!\n')
```

- Ler de um arquivo de texto (`txt`):

```python
with open('dados.txt', 'r') as arquivo:
    conteudo = arquivo.read()
    print(conteudo)  # Saída: Este é um exemplo de texto.\nPython é uma linguagem poderosa!\n
```

### 2. Arquivos CSV
Os arquivos CSV (Comma-Separated Values) são comumente usados para armazenar dados tabulares em um formato simples e legível.

- Escrever em um arquivo CSV:

```python
import csv

dados = [
    ['Nome', 'Idade', 'Cidade'],
    ['Alice', 25, 'São Paulo'],
    ['Bob', 30, 'Rio de Janeiro'],
    ['Carol', 28, 'Belo Horizonte']
]

with open('dados.csv', 'w', newline='') as arquivo:
    escritor_csv = csv.writer(arquivo)
    escritor_csv.writerows(dados)
```

Usamos o módulo csv para escrever dados em um formato tabular. O método writerows() escreve múltiplas linhas de dados no arquivo CSV.

- Ler de um arquivo CSV:

```python
with open('dados.csv', 'r', newline='') as arquivo:
    leitor_csv = csv.reader(arquivo)
    for linha in leitor_csv:
        print(linha)  # Saída: Cada linha do arquivo CSV será impressa como uma lista de strings
```

O método reader() do módulo csv lê as linhas do arquivo CSV, cada uma como uma lista de strings, representando as colunas.

### 3. Arquivos JSON
JSON (JavaScript Object Notation) é um formato leve de troca de dados amplamente utilizado por sua simplicidade e legibilidade.

- Escrever em um arquivo JSON

```python
import json

dados_json = {
    'nome': 'Alice',
    'idade': 30,
    'cidade': 'São Paulo'
}

with open('dados.json', 'w') as arquivo:
    json.dump(dados_json, arquivo)
```
