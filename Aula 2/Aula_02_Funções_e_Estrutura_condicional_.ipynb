### Funções na Matemática e na Programação

As funções são um conceito fundamental tanto na matemática quanto na programação, embora elas sejam usadas de maneiras um pouco diferentes em cada campo. Vamos explorar a função \( f(x) = x + 1 \) para entender melhor essa relação.

#### Funções na Matemática

Na matemática, uma função é uma relação entre um conjunto de entradas (domínio) e um conjunto de saídas (imagem), onde cada entrada está associada a exatamente uma saída. Por exemplo, a função \( f(x) = x + 1 \) toma um número \( x \) e retorna \( x + 1 \).

**Exemplo:**

- Se \( x = 2 \), então \( f(2) = 2 + 1 = 3 \).
- Se \( x = 5 \), então \( f(5) = 5 + 1 = 6 \).

Essa função é uma regra que associa cada número \( x \) ao número \( x + 1 \).

#### Funções na Programação

Na programação, uma função é um bloco de código reutilizável que realiza uma tarefa específica. Uma função pode receber entradas (também chamadas de argumentos ou parâmetros) e pode retornar uma saída. A ideia é semelhante à da matemática, mas com mais flexibilidade e aplicabilidade prática.

Vamos ver como podemos implementar a função \( f(x) = x + 1 \) em algumas linguagens de programação.

##### Python

### Escopo de uma Função em Python

O escopo de uma função em Python refere-se à área do programa onde as variáveis definidas podem ser acessadas. Entender o escopo é crucial para evitar erros e escrever código eficiente e organizado.

#### Tipos de Escopo

1. **Escopo Local:** Variáveis definidas dentro de uma função só podem ser acessadas dentro dessa função.
2. **Escopo Global:** Variáveis definidas fora de qualquer função podem ser acessadas em qualquer parte do programa.
3. **Escopo Encarregado:** Variáveis definidas dentro de uma função que chama outra função.
4. **Escopo Interno:** Variáveis definidas dentro de uma função aninhada.

### Escopo Local

Variáveis definidas dentro de uma função estão no escopo local e não podem ser acessadas fora dessa função.




````python
def minha_funcao():
    x = 10  # Escopo local
    print(x)

minha_funcao()
print(x)
````

## Escopo Global
Variáveis definidas fora de qualquer função estão no escopo global e podem ser acessadas de qualquer lugar no programa.

````python
y = 20  # Escopo global

def outra_funcao():
    print(y)

outra_funcao()
print(y)
````

````python
z = 30  # Escopo global

def modificar_global():
    global z
    z = 40

print(z)
modificar_global()
print(z)
````

## Escopo Encarregado e Interno
Funções podem ser aninhadas, e variáveis definidas em uma função interna têm escopo local apenas dentro dessa função.

````python
def externa():
    a = 50  # Escopo local na função 'externa'

    def interna():
        b = 60  # Escopo local na função 'interna'
        print(a)  # Pode acessar 'a' da função externa
        print(b)

    interna()
    print(a)
    # print(b)  # Erro: NameError: name 'b' is not defined

externa()
````

##Dicas para Gerenciar o Escopo
* Evite usar muitas variáveis globais: Elas podem ser modificadas de qualquer lugar, o que pode causar bugs difíceis de encontrar.
* Prefira variáveis locais: Elas ajudam a manter o código modular e fácil de entender.
* Use nomes claros e descritivos: Para evitar confusão entre variáveis globais e locais.

##Indentação em Python
Indentação refere-se ao uso de espaços ou tabulações no início de uma linha de código. Em Python, a indentação não é apenas uma questão de estilo, mas uma parte fundamental da sintaxe da linguagem. A indentação é usada para definir blocos de código, como o corpo de funções, loops, condicionais, etc.

##Regras de Indentação
* Consistência: Use a mesma quantidade de espaços ou tabulações em todo o seu código. O padrão é usar 4 espaços para cada nível de indentação.
* Obrigatoriedade: Python exige que blocos de código sejam indentados de maneira consistente. Caso contrário, você encontrará erros de indentação.

1. if <br>
A declaração if é utilizada para verificar uma condição simples e executar um bloco de código se essa condição for verdadeira.

````python
numero = 10

if numero % 2 == 0:
    print(f"{numero} é um número par.")
````

2. else<br>
A declaração else é usada em conjunto com if para executar um bloco de código caso a condição do if seja falsa.

````python
idade = 17

if idade >= 18:
    print("Você é maior de idade.")
else:
    print("Você é menor de idade.")
````

3. elif <br>
A declaração elif é uma abreviação de "else if" e é utilizada para verificar múltiplas condições encadeadas. Ela vem após um bloco if e antes de um possível bloco else.

````python
idade = 25

if idade < 12:
    print("Criança")
if idade < 18:
    print("Adolescente")
if idade < 60:
    print("Adulto")
else:
    print("Idoso")
````

# Diferença entre if e elif
* if : verifica sempre
* elif : verifica até o primeiro verdadeiro

````python
nota = 75

if nota >= 90:
    print("A nota é A.")
if nota >= 80 and nota < 90:
    print("A nota é B.")
if nota >= 70 and nota < 80:
    print("A nota é C.")
if nota >= 60 and nota < 70:
    print("A nota é D.")
if nota < 60:
    print("A nota é F.")
````

````python
nota = 75

if nota >= 90:
    print("A nota é A.")
elif nota >= 80:
    print("A nota é B.")
elif nota >= 70:
    print("A nota é C.")
elif nota >= 60:
    print("A nota é D.")
else:
    print("A nota é F.")
````

#Boas Práticas para o uso de estruturas condicionais

As estruturas condicionais são fundamentais para o controle do fluxo de um programa. Em Python, as principais estruturas condicionais são `if`, `elif` e `else`. Seguir boas práticas ao usá-las ajuda a tornar o código mais legível, eficiente e fácil de manter. Aqui estão algumas dicas:

## 1. Clareza e Legibilidade

### Use Indentação Adequada
Python depende da indentação para definir blocos de código. Certifique-se de usar a mesma quantidade de espaços (ou tabulações) consistentemente.

```python
if condição:
    # bloco de código
    fazer_algo()
else:
    # outro bloco de código
    fazer_algo_diferente()


# Nomes de Variáveis Significativos
Use nomes de variáveis que sejam claros e descritivos para que o propósito da condição seja evidente.

````python
idade = 20
if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
````

# Simplificação de Condições
Evite Condições Aninhadas Desnecessárias
Condições aninhadas podem tornar o código difícil de seguir. Sempre que possível, simplifique o fluxo lógico.

````python
# Evite
if a:
    if b:
        fazer_algo()

# Prefira
if a and b:
    fazer_algo()
````

# Utilize Expressões Booleanas
Se a condição é uma expressão booleana, não é necessário compará-la explicitamente a True ou False.

````python
# Evite
if ativo == True:
    fazer_algo()

# Prefira
if ativo:
    fazer_algo()
````

# Evite Condições Redundantes
Não Repita Condições Desnecessárias
Verifique se as condições são mínimas e suficientes para cobrir todos os casos.

````python
# Evite
if x > 10:
    print("Maior que 10")
elif x > 5:
    print("Maior que 5, mas menor ou igual a 10")

# Prefira
if x > 10:
    print("Maior que 10")
elif x > 5:
    print("Maior que 5")
````
