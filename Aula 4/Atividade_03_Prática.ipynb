As operações com matrizes são essenciais em áreas como álgebra linear, gráficos computacionais, inteligência artificial e processamento de imagens. Nesta questão, você deverá implementar operações com matrizes em Python que envolvem desde as operações básicas até as avançadas.

## Definições e Conceitos:

1. **Adição de Matrizes**: Duas matrizes podem ser somadas se tiverem as mesmas dimensões. A soma de duas matrizes \( A \) e \( B \) é obtida somando os elementos correspondentes de cada matriz. Ou seja, para cada \( i \) e \( j \),
   \[
   (A + B)_{ij} = A_{ij} + B_{ij}
   \]
   
2. **Subtração de Matrizes**: Assim como a adição, a subtração de duas matrizes \( A \) e \( B \) ocorre elemento a elemento, desde que as matrizes tenham as mesmas dimensões.
   \[
   (A - B)_{ij} = A_{ij} - B_{ij}
   \]
   
3. **Multiplicação de Matrizes**: A multiplicação de duas matrizes \( A \) e \( B \) é possível quando o número de colunas de \( A \) é igual ao número de linhas de \( B \). O elemento \( C_{ij} \) da matriz resultante \( C \) é calculado pela multiplicação escalar da linha \( i \) de \( A \) pela coluna \( j \) de \( B \).
   \[
   C_{ij} = \sum_{k} A_{ik} \times B_{kj}
   \]
   
4. **Determinante**: O determinante de uma matriz quadrada \( A \) é um número que dá informações sobre várias propriedades da matriz, como se ela é invertível ou não. Para uma matriz nxn:
   \[
   A = \begin{bmatrix} a & b \\ c & d \end{bmatrix}, \quad det(A) = ad - bc
   \]

5. **Matriz Inversa**: A inversa de uma matriz \( A \), denotada como \( A^{-1} \), é uma matriz tal que:
   \[
   A \times A^{-1} = I
   \]
   onde \( I \) é a matriz identidade. A matriz inversa só existe se o determinante de \( A \) for diferente de zero.

6. **Escalonamento**: O escalonamento de uma matriz consiste em aplicar operações elementares para transformá-la na forma escalonada por linhas, onde cada linha começa com um número maior de zeros que a linha anterior.

7. **Matriz Transposta**: A transposta de uma matriz \( A \) é obtida trocando suas linhas pelas colunas. A transposta de \( A \) é denotada como \( A^T \), e se:
   \[
   A = \begin{bmatrix} a & b \\ c & d \end{bmatrix}, \quad A^T = \begin{bmatrix} a & c \\ b & d \end{bmatrix}
   \]

### Conceitos a Serem Implementados

1. **Busca Linear**
   - **Conceito:** Método de busca que percorre a lista do início ao fim, comparando cada elemento com o valor desejado.
   - **Complexidade de Tempo:** O(n)
   - **Quando Usar:** Listas pequenas ou não ordenadas.

2. **Busca Binária**
   - **Conceito:** Método eficiente para encontrar um item em uma lista ordenada, dividindo a lista em metades a cada iteração.
   - **Complexidade de Tempo:** O(log n)
   - **Quando Usar:** Listas grandes e ordenadas.

3. **Inversão de Lista**
   - **Conceito:** Reverte a ordem dos elementos em uma lista.
   - **Complexidade de Tempo:** O(n)
   - **Quando Usar:** Para manipular a ordem dos elementos.

4. **Bubble Sort**
   - **Conceito:** Algoritmo de ordenação simples que compara e troca pares adjacentes de elementos até que a lista esteja ordenada.
   - **Complexidade de Tempo:** O(n²) no pior caso
   - **Quando Usar:** Para listas pequenas ou quando a simplicidade é mais importante que a eficiência.

5. **Selection Sort**
   - **Conceito:** Algoritmo de ordenação que encontra o menor elemento da parte não ordenada e o coloca na parte ordenada.
   - **Complexidade de Tempo:** O(n²)
   - **Quando Usar:** Listas pequenas ou quando a simplicidade é mais importante que a eficiência.
