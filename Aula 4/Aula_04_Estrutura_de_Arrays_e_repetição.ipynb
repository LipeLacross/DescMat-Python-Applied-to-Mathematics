# O que é uma lista?

Em Python, uma lista é uma estrutura de dados que permite armazenar e manipular um conjunto de dados. A lista pode ser armazenada em uma variável, e a partir dela, há um conjunto de métodos e operações que podem ser realizados com os dados armazenados. Sendo possível inserir diferentes tipos de dados, como Strings, números inteiros e flutuantes, isto para o Python.

# Abstração de uma Lista

Apesar de ser bem sugestivo é possível que não seja tão claro a primeira vista. Então para visualizar melhor o que seria uma lista, criemos um cenário hipotético.

Imagine que algum familiar seu pediu para que você fosse ao mercado e para que nada fosse esquecido você decidiu anotar os itens que irá comprar. A princípio você tem um papel com o título "Produtos para Comprar" e irá adicionar um abaixo do outro os produtos que você precisa comprar à lista. Ficando algo mais o menos assim:

**Produtos para comprar**

1.   Arroz
2.   Feijão
3. Farinha
4. ...

Listas em python seguem a mesma lógica que uma lista da vida real, onde você pode inserir, remover, alterar ou mover, os itens que você deseja.

# Propriedades das listas



1.   **Mutáveis:** Ao contrário das tuplas, as listas em Python são mutáveis, o que significa que você pode alterar seus elementos após a criação.
2.   **Ordenadas:** Os elementos de uma lista mantêm a ordem na qual são adicionados. Cada elemento pode ser acessado por um índice, começando do zero.
3.   **Tipos Mistos:** Uma lista pode conter elementos de diferentes tipos de dados, incluindo inteiros, strings, objetos, outras listas, etc.
4.   **Dinâmicas:** As listas podem crescer ou diminuir de tamanho conforme necessário. Você pode adicionar, remover ou alterar elementos.

Na nossa lista podemos observar uma das propriedades, a dos tipos mistos, onde temos uma lista contendo, números inteiros, strings e até mesmo outra lista.

# Operações em listas
**Acessando um elemento da lista.**

Para acessarmos um elemento, é importante termos a noção do conceito de índices de uma lista. Em termos simples, o índice se refere a posição do elemento na lista. Como assim? Bem digamos que eu tenha a seguinte lista:

`frutas = ["maça", "laranja", "limão"]`

Ao total eu tenho 3 itens em minha lista, e seus ìndices serão respectivamente:

*   frutas[0] = "maça"
*   frutas[1] = "laranja"
* frutas[2] = "limão"

Onde o valor entre os colchetes representa o índice e posição do meu item na lista. Você pode observar que iniciamos do zero, e está absolutamente correto. O índice de nossas listas irão se iniciar a partir do índice '0', fixe bem esse conceito, pois ao operarmos em lista temos que ter em mente ele. A motivação disto está atrelada a razões técnicas e históricas sendo uma convenção, mas caso queira entender a fundo basta acessar o material extra.

````python
# acessando um item da minha lista

lista = [1, 2, 3, 4, 5]
print(lista[0])  #1
print(lista[1])  #2
print(lista[-1]) #5
````

````python
# modificando elementos da minha lista

lista = [1, 2, 3, 4, 5]
lista[0] = 10
print(lista)
````

````python
# slicing

comidas = ["espaguete", "pizza", "lasanha", "pão", "feijoada"]
sub_lista = comidas[1:3]
print(sub_lista)
````

````python
# adicionando elementos a minha lista

mensagens = ["olá", "oi", "bem-vindo"]
mensagens.append("hola")  # Adiciona ao final da lista
mensagens.extend(["good", "bad", "python"])  # Adiciona múltiplos elementos
nova_lista = mensagens + [10, 11]  # Cria uma nova lista

print(mensagens)
print(nova_lista)
````

````python
# cuidados a serem tomados
lista_a = [1,2,3]
lista_b = lista_a
lista_a.append(47)
print("lista_a", lista_a)
print("lista_b", lista_b)

lista_b.append(99)
print("lista_a", lista_a)
print("lista_b", lista_b)

# Com lista_a=lista_b, teremos que as duas listas serão a mesma

# como copiar uma lista
a=[1,2,3]
b=a[:]
a.append(47)
print("a", a)
print("b", b)
````

````python
# removendo elementos da minha lista

minha_lista = [1, 2, 3, 'a', 'b', 'c', [4, 5, 6]]

minha_lista.remove('a')                 # Remove o primeiro elemento que é igual a 'a'
elemento_removido = minha_lista.pop(2)  # Remove e retorna o elemento no índice 2
del minha_lista[0]                      # Remove o elemento no índice 0

print(minha_lista)
print(elemento_removido)
````

#Métodos Úteis
As listas em Python vêm com vários métodos embutidos que facilitam a manipulação dos dados:

**append(x):** Adiciona o item x ao final da lista.

**extend(iterável):** Adiciona todos os itens de um iterável à lista.

**insert(i, x):** Insere o item x na posição i.

**remove(x):** Remove a primeira ocorrência do item x.

**pop([i]):** Remove o item na posição i e o retorna. Se i não for especificado,
remove e retorna o último item.

**clear():** Remove todos os itens da lista.

**index(x[, start[, end]]):** Retorna o índice da primeira ocorrência do item x.

**count(x):** Retorna o número de vezes que o item x aparece na lista.

**sort(key=None, reverse=False):** Ordena os itens da lista.

**reverse():** Inverte a ordem dos itens na lista.

**copy():** Retorna uma cópia superficial da lista.

Tuplas em Python são coleções ordenadas e imutáveis de elementos. Elas são semelhantes às listas, mas com a diferença de que, uma vez criadas, não podem ser alteradas, ou seja, não é possível adicionar, remover ou modificar elementos. A sintaxe básica para criar uma tupla é usar parênteses ().

**Tupla vazia**

````python
tupla = ()
````

**Tupla com elementos**

````python
tupla = (1, 2, 3)
````

**Tupla sem parênteses (Empacotamento de Tupla)**

# **Estruturas de repetição**

Em Python, as estruturas de repetição permitem executar um bloco de código várias vezes. Existem duas principais estruturas de repetição: for e while.

# For

A estrutura for é usada para iterar sobre uma sequência, como uma lista, tupla, dicionário, conjunto ou string. Onde podemos delimitar o número de vezes a qual a estrutura irá iterar, tornando-a uma escolha favorável quando se sabe o número de iterações necessárias.

**Sintaxe básica de interação por número fixo**

````python
for i in range(5):
  print(i)
````

**Iterando sobre uma lista**

````python
cesta_de_frutas = ["maça", "laranja", "limão"]

for fruta in cesta_de_frutas:
  print(fruta)
````

**Iterando sobre uma String**

````python
palavra = "Descmat"

for letra in palavra:
  print(letra)
````

# While

A estrutura while repete um bloco de código enquanto uma condição é verdadeira. Sendo interessante utilizá-lô quando temos uma condição de parada, porém não sabemos quantas iterações são necessárias para chegar nessa condição.

````python
contador = 0

while contador < 5:
  print(contador)
  contador += 1 # ou contador = contador + 1. Garante que nosso programa não entre em um loop infinito.
````

# Palavras chaves atuantes em estruturas de repetição


* **break:** Encerra o loop imediatamente.
* **continue:** Pula para a próxima iteração do loop.
* **else:** Executa um bloco de código quando o loop termina sem ser interrompido por um break.



**break**

````python
for i in range(10):
  if i == 5:
    break
  print(i)
````

**continue**

````python
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
````

**else**

````python
for i in range(5):
    print(i)
else:
    print("Loop concluído sem interrupções.")
````

````python
A = [[1,2,3],
     [3,2,1],
     [3,3,3]]



for i in range(len(A)):
  for j in range(len(A[i])):
    print(A[i][j])
````
